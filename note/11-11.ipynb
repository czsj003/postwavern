第二步：手写第一个 HTTP 接口和测试

#### 本阶段目标
实现：

GET /api/health

```
响应：
{
  "status": "UP"
}
```

调用过程是：
```
浏览器
→ Tomcat
→ DispatcherServlet
→ HealthController
→ HealthResponse
→ JSON
```


#### HealthController
@RestController：告诉 Spring，这个类负责处理 HTTP 请求，返回值写进响应体。

@RequestMapping("/api/health")：定义这个 Controller 的基础路径。

@GetMapping：把 HTTP GET 请求映射到这个方法。

new HealthResponse("UP")：返回 Java 对象。

Spring 使用 JSON 转换器把对象变成 {"status":"UP"}。

因为 HealthController 位于 com.postwave.backend 的子包中，所以启动类的组件扫描能够自动发现它。

#### HealthControllerTest

backend/src/test/java/com/postwave/backend/health/HealthControllerTest.java

理解这个测试

@WebMvcTest(HealthController.class)：
- 只加载 Web MVC 相关组件
- 只测试指定 Controller
- 不启动真实 Tomcat
- 不加载完整应用
- 执行速度比完整集成测试快
MockMvc 模拟一次 HTTP 请求：
mockMvc.perform(get("/api/health"))
